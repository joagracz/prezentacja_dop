## Typowanie dynamiczne

Python jest językiem programowania o typowaniu dynamicznym. Oznacza to, że interpreter Pythona sprawdza poprawność typów danych dopiero podczas wykonywania programu. Co więcej, typ zmiennej może zmieniać się w trakcie działania programu.

```python
if False:
    print(1 + "two") # Ten fragment kodu nie zostanie wykonany, dlatego nie pojawia się TypeError
else:
    print(1 + 2)
```

```python
# tu już jest wykonywany, dlatego pojawia się błąd
print(1 + "two") 
```

```python
# zmiana typu zmiennej
name = "Anna"
type(name)
```

```python
name = 28.1
type(name)
```

## Typowanie statyczne

Przeciwieństwem typowania dynamicznego jest typowanie statyczne. W tym przypadku kontrola poprawności typów danych odbywa się bez konieczności uruchamiania programu. W wielu językach programowania, takich jak C czy Java, sprawdzanie typów następuje podczas kompilacji kodu.
W językach wykorzystujących typowanie statyczne zmienne mają zazwyczaj określony typ, którego nie można dowolnie zmieniać.

```python
# przykład w języku Java
String name; # deklarujemy zmienną name i określamy, że będzie ona przechowywać obiekty typu string
name = "Anna"; # przypisujemy wartość
name = 28.1f; # błąd, ponieważ typ przypisanej wartości jest niezgodny z typem zmiennej
```

## Duck Typing

Nazwa tego podejścia pochodzi od popularnego powiedzenia „Jeśli coś chodzi jak kaczka i kwacze jak kaczka, to musi być kaczką”. W programowaniu oznacza to, że ważniejsza od konkretnego typu lub klasy obiektu jest jego funkcjonalność, czyli metody i atrybuty, które udostępnia.

```python
class MyBook:
    def __len__(self):
        return 95022

my_book = MyBook()
len(my_book)
```

W powyższym przykładzie tworzymy klasę MyBook, która zawiera metodę *\_\_len\_\_()*. Metoda ta zwraca wartość 95022. Dzięki temu możemy użyć funkcji *len()* na obiekcie klasy MyBook, mimo że nie jest on standardowym typem danych, takim jak lista czy tekst. Ogólnie, aby można było wywołać funkcję *len(obj)*, obiekt musi udostępniać metodę *\_\_len\_\_()*. Nie ma przy tym znaczenia typ ani klasa obiektu.

## Adnotacje typów

Python pozostaje językiem o typowaniu dynamicznym. Adnotacje typów umożliwiają jednak stosowanie statycznego sprawdzania typów na etapie tworzenia kodu. Pozwalają określić, jakiego typu danych oczekuje dana zmienna, funkcja lub jej argument. Jednak same z siebie nie powodują automatycznego wymuszania zgodności typów przez interpreter Pythona. Są jedynie wskazówką dla programisty.

```python
def headline(text: str, align: bool = True) -> str:
    if align:
        return f"{text.title()}\n{'-' * len(text)}"
    else:
        return f" {text.title()} ".center(50, "o")
```

```python
print(headline("python type checking"))
```

```python
print(headline("python type checking", align=False))
```

Mimo zadeklarowania parametru *align* jako typu *bool*, przekazanie mu wartości typu *str* nie powoduje błędu podczas wykonywania programu.

```python
print(headline("python type checking", align = "left")) 
```

Chociaż adnotacje typów same w sobie nie wymuszają zgodności typów podczas wykonywania programu, to mogą być wykorzystywane przez zewnętrzne narzędzia do statycznego sprawdzania poprawności typów. Do najpopularniejszych z nich należą:
- mypy – narzędzie do statycznej analizy typów, które wykorzystuje adnotacje typów do wykrywania potencjalnych błędów
- PyCharm – ostrzega o niezgodnościach typów już podczas pisania programu

W przypadku zmiennych również możemy stosować adnotacje typów:

```python
pi: float = 3.142
print(pi)
```

Informacje o adnotacjach funkcji są dostępne przez \_\_annotations\_\_.

```python
headline.__annotations__
```

## Type Comments

Adnotacje typów zostały wprowadzone w Pythonie 3 i nie zostały przeniesione do Pythona 2. Nie możemy z nich więc korzystać podczas tworzenia kodu, który musi być kompatybilny ze starszymi wersjami Pythona. Alternatywnym rozwiązaniem są komentarze dotyczące typów (type comments). Są to specjalnie sformatowane komentarze, które umożliwiają dodawanie adnotacji typów, a jednocześnie zachowują kompatybilność ze starszym kodem.

```python
def headline(text, width=80, fill_char="-"):
    # type: (str, int, str) -> str
    return f" {text.title()} ".center(width, fill_char)

print(headline("type comments work", width=40))
```

```python
def headline(
    text,           # type: str
    width=80,       # type: int
    fill_char="-",  # type: str
):                  # type: (...) -> str
    return f" {text.title()} ".center(width, fill_char)

print(headline("type comments work", width=40))
```

```python
pi = 3.142  # type: float
```

Zastosowanie **from \_\_future\_\_ import annotations** zmienia sposób, w jaki Python interpretuje adnotacje typów. Domyślnie Python odczytuje je w momencie definiowania funkcji lub klasy, więc wykorzystywane typy muszą być już wcześniej zdefiniowane. Po dodaniu tej instrukcji Python nie interpretuje adnotacji od razu, lecz przechowuje je w postaci łańcuchów znaków (*str*). Dzięki temu można używać nazw klas w adnotacjach, nawet jeśli ich definicje znajdują się niżej w kodzie. 

```python
from __future__ import annotations

def create_student() -> Student:
    return Student()

class Student:
    pass
```

## Zalety i wady stosowania adnotacji typów

Zalety:
- Wykrywanie błędów – umożliwiają wykrywanie niektórych błędów już na etapie analizy kodu, zanim program zostanie uruchomiony.
- Lepsza dokumentacja kodu – określają oczekiwane typy argumentów funkcji i zwracanych wartości, co zwiększa czytelność kodu.
- Usprawnienie pracy w środowiskach programistycznych i narzędziach do analizy kodu – adnotacje typów ułatwiają analizę kodu, dzięki czemu środowiska programistyczne mogą oferować dokładniejsze podpowiedzi, automatyczne uzupełnianie kodu i wykrywanie potencjalnych błędów.

Wady:
- Większy nakład czasu i pracy programisty – dodawanie adnotacji typów wymaga dodatkowego czasu i wysiłku podczas pisania kodu, chociaż może ograniczyć czas poświęcany później na wyszukiwanie i naprawianie błędów.
- Najlepsza współpraca z nowszymi wersjami Pythona – adnotacje typów wprowadzono w Pythonie 3.0. W starszej wersji Pythona 2.7 można korzystać z komentarzy dotyczących typów (type comments), jednak nowsze rozwiązania, takie jak adnotacje zmiennych i odroczone obliczanie adnotacji, zapewniają wygodniejszą pracę z typami w Pythonie 3.6 i 3.7 oraz nowszych wersjach.
